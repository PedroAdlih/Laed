```python
#Tipo Abstrato de Dados da Lista Duplamente Encadeada
class No:
    def __init__(self, valor):
        self.valor = valor
        self.prox = None
        self.ant = None

class Lista:
    def __init__(self):
        self.inicio = None
        self.fim = None

    def inserir(self, valor):
        novo = No(valor)
        if self.inicio is None:
            self.inicio = novo
            self.fim = novo
        else:
            self.fim.prox = novo
            novo.ant = self.fim
            self.fim = novo

    def imprimir(self):
        head = self.inicio
        while head is not None:
            print(head.valor, end=" -> ")
            head = head.prox
        print("None")
```

```python
#Questão 1
def elementoCentral(p):
    if p is None:
        return None
    lento = p
    rapido = p
    while rapido is not None and rapido.prox is not None:
        lento = lento.prox
        rapido = rapido.prox.prox
    return lento
#B) Como rapido percorre a lista n vezes e lento ate n vezes
#temos O(n) a Complexidade
```

```python
#Questão 2
def atualizar(p, x, y):
    head = p
    alvo = None
    while head is not None:
      if head.valor == x:
        alvo = head
        break
      head = head.prox
    if alvo is None:
        return p
    alvo.valor = y
    if alvo.ant is not None:
        alvo.ant.prox = alvo.prox
    else:
        p = alvo.prox
    if alvo.prox is not None:
        alvo.prox.ant = alvo.ant
    alvo.prox = None
    alvo.ant = None
    if p is None:
        p = alvo
        return p
    if y <= p.valor:
        alvo.prox = p
        p.ant = alvo
        p = alvo
        return p
    atual = p
    while atual.prox is not None and atual.prox.valor < y:
        atual = atual.prox
    alvo.prox = atual.prox
    if atual.prox is not None:
        atual.prox.ant = alvo
    atual.prox = alvo
    alvo.ant = atual
    return p
#B)Complexidade O(n), pois no pior caso percorre para encontrar x
# e depois para encontrar y
```

```python
#Questão 3
def trocar(p, q):
    r = q.prox
    antQ = q.ant
    proxR = r.prox
    if antQ is not None:
        antQ.prox = r
    else:
        p = r
    r.ant = antQ
    q.prox = proxR
    if proxR is not None:
        proxR.ant = q
    r.prox = q
    q.ant = r
    return p

def umaVarredura(p):
    trocou = False
    q = p
    while q is not None and q.prox is not None:
        if q.valor > q.prox.valor:
            p = trocar(p, q)
            trocou = True
        q = q.prox  # apos a troca, q segue para o elemento que ficou na sua frente
    return p, trocou

def ordenar(p):
    trocou = True
    while trocou:
        p, trocou = umaVarredura(p)
    return p

#B) Complexidade O(n^2), ja que podemos ter n varreduras
```

```python
#Questão 4
def particao(p, r, k):
    q = p
    while True:
        while q is not None and q.valor <= k and q != r:
            q = q.prox
        while r is not None and r.valor > k and q != r:
            r = r.ant
        if q is None or r is None or q == r or q.ant == r:
            break
        q.valor, r.valor = r.valor, q.valor
        q = q.prox
    return p

#B)Ja que q e r percorrem a lista uma unica vez temos O(n)
```

```python
#Questão 5
class NoEsparso:
    def __init__(self, valor, indice):
        self.valor = valor
        self.indice = indice
        self.prox = None

def construirVetorEsparso(vetor):
    p = None
    fim = None
    for i in range(len(vetor)):
        if vetor[i] != 0:
            novo = NoEsparso(vetor[i], i)
            if p is None:
                p = novo
            else:
                fim.prox = novo
            fim = novo
    return p

#B) O(n), ja que pervorre o vetor uma unica vez
```

```python
#Questão 6
def buscaPorIndice(p, k):
    head = p
    while head is not None:
        if head.indice == k:
            return head.valor
        head = head.prox
    return 0  # posicoes nao armazenadas valem 0

def buscaPorValor(p, x):
    head = p
    while head is not None:
        if head.valor == x:
            return head.indice
        head = head.prox
    return -1

def atualizacaoEsparso(p, x, k):
    head = p
    while head is not None:
        if head.indice == k:
            head.valor = x
            return p
        head = head.prox
    # posicao k ainda nao existia na lista: se x != 0, insere um novo no
    if x != 0:
        novo = NoEsparso(x, k)
        if p is None or k < p.indice:
            novo.prox = p
            p = novo
        else:
            atual = p
            while atual.prox is not None and atual.prox.indice < k:
                atual = atual.prox
            novo.prox = atual.prox
            atual.prox = novo
    return p

#B) Complexidade: O(n)
```

```python
#Questão 7
def transformarEmListaDeListas(p, k):
    L = []
    atual = p
    while atual is not None:
        subInicio = atual
        subInicio.ant = None
        subFim = atual
        cont = 1
        while cont < k and atual.prox is not None:
            atual = atual.prox
            subFim = atual
            cont += 1
        proximo = atual.prox
        subFim.prox = None
        if proximo is not None:
            proximo.ant = None
        L.append(subInicio)
        atual = proximo
    return L

#B) O(n), pois cada no é visitado apenas uma vez
```

```python
#Questão 10
def buscaListaDeListas(L, x):
    for i, sub in enumerate(L):
        proximaExiste = i + 1 < len(L)
        if not proximaExiste or (L[i + 1] is not None and x < L[i + 1].valor):
            head = sub
            while head is not None:
                if head.valor == x:
                    return head
                head = head.prox
            return None
    return None

def insercaoListaDeListas(L, x):
    for i, sub in enumerate(L):
        proximaExiste = i + 1 < len(L)
        if not proximaExiste or (L[i + 1] is not None and x < L[i + 1].valor):
            novo = No(x)
            head = sub
            if head is None or x <= head.valor:
                novo.prox = head
                if head is not None:
                    head.ant = novo
                L[i] = novo
                return L
            while head.prox is not None and head.prox.valor < x:
                head = head.prox
            novo.prox = head.prox
            if head.prox is not None:
                head.prox.ant = novo
            head.prox = novo
            novo.ant = head
            return L
    return L

def remocaoListaDeListas(L, x):
    for i, sub in enumerate(L):
        head = sub
        while head is not None:
            if head.valor == x:
                if head.ant is not None:
                    head.ant.prox = head.prox
                else:
                    L[i] = head.prox
                if head.prox is not None:
                    head.prox.ant = head.ant
                return L
            head = head.prox
    return L

#B) O(k), tendo k o tamanho da sublista
```
