```python
#Tipo Abstrato de Dados da Lista Encadeada
class No:
    def __init__(self, valor):
        self.valor = valor
        self.prox = None

class Lista:
    def __init__(self):
        self.inicio = None

    def inserir(self, valor):
        novo = No(valor)
        if self.inicio is None:
            self.inicio = novo
        else:
            temp = self.inicio
            while temp.prox is not None:
                temp = temp.prox
            temp.prox = novo

    def imprimir(self):
        head = self.inicio
        while head is not None:
            print(head.valor, end=" -> ")
            head = head.prox
        print("None")
```

```python
#Questão 1
def maiorNoFinal(p):
    if p is None:
        return p
    maior = p
    ant_maior = None
    ant = None
    head = p
    # Encontra o maior
    while head is not None:
        if head.valor > maior.valor:
            maior = head
            ant_maior = ant
        ant = head
        head = head.prox
    # Se o maior já tá no final
    if maior.prox is None:
        return p
    # Retirar o maior da posição atual
    if ant_maior is None:
        p = maior.prox
    else:
        ant_maior.prox = maior.prox
    # Percorrer até o último elemento
    head = p
    while head.prox is not None:
        head = head.prox
    # Colocar o maior no final
    head.prox = maior
    maior.prox = None
    return p

#B) O algoritmo é O(n)
```

```python
#Questão 2
def separarParesImpares(p):
    pImpar = None
    pPar = None
    fimImpar = None
    fimPar = None
    head = p
    while head is not None:
        prox = head.prox
        # parte ímpar
        if head.valor % 2 != 0:
            if pImpar is None:
                pImpar = head
                fimImpar = head
            else:
                fimImpar.prox = head
                fimImpar = head
        # parte par
        else:
            if pPar is None:
                pPar = head
                fimPar = head
            else:
                fimPar.prox = head
                fimPar = head
        head = prox
    if fimImpar is not None:
        fimImpar.prox = None
    if fimPar is not None:
        fimPar.prox = None
    return pImpar, pPar

# B) A complexidade desse algoritmo é O(n) já que o percurso é feito apenas
# uma vez na lista encadeada
```

```python
#Questão 3
def inverter(p):
  ant = None
  head = p
  while head is not None:
    prox = head.prox
    head.prox = ant
    ant = head
    head = prox
  return ant
#B) O laço percorre uma única vez a lista encadeada
# logo, o tempo de execução éO(n)
```

```python
#Questão 4
def duplicarImpares(p):
  head = p
  while head is not None:
    if head.valor % 2 != 0:
      novo = No(head.valor)
      novo.prox = head.prox
      head.prox = novo
      head = novo.prox
    else:
      head = head.prox
  return p

#B) A complexidade do algoritmo é O(n) ja que elaprecisa
#percorrer aapenas uma vez a lista
```

```python
#Questão 5
def intercalar(p1, p2):
  if p1 is None:
    return p2
  if p2 is None:
    return p1
  if p1.valor < p2.valor:
    p = p1
    p1 = p1.prox
  else:
    p = p2
    p2 = p2.prox
  fim = p
  while p1 is not None and p2 is not None:
    if p1.valor < p2.valor:
      fim.prox = p1
      p1 = p1.prox
    else:
      fim.prox = p2
      p2 = p2.prox
      fim = fim.prox
  if p1 is not None:
    fim.prox = p1
  else:
    fim.prox = p2
  return p
#B) Tendo uma lista de tamanho n e outra de tamanho m
#,etendo um algoritmo percorrendo a lista apenas uma vez
# temos: O(n+m)
```

```python
#Questão 6
def particionar(p, k):
  pMenor = None
  fimMenor = None
  pMaior = None
  fimMaior = None
  head = p
  while head is not None:
    prox = head.prox
    if head.valor <= k:
      if pMenor is None:
        pMenor = head
        fimMenor = head
      else:
        fimMenor.prox = head
        fimMenor = head
    else:
      if pMaior is None:
        pMaior = head
        fimMaior = head
      else:
        fimMaior.prox = head
        fimMaior = head
    head = prox
  if pMenor is None:
    return pMaior
  fimMenor.prox = pMaior
  if fimMaior is not None:
    fimMaior.prox = None
  return pMenor
#B)Complexidade O(n), porque a lista é percorrida uma vez
```

```python
#Questão 7
def removerCopias(p, k):
  while p is not None and p.valor == k:
    p = p.prox
  if p is None:
    return p
  ant = p
  head = p.prox
  while head is not None:
    if head.valor == k:
      ant.prox = head.prox
      head = ant.prox
    else:
      ant = head
      head = head.prox
  return p

#B)Complexidade O(n) ja que percorre a lista uma vez
```

```python
#Questão 8
def maisFrequente(p):
  if p is None:
    return None, 0
  maior = p.valor
  maiorCont = 0
  head = p
  while head is not None:
    cont = 0
    aux = p
    while aux is not None:
      if aux.valor == head.valor:
        cont += 1
      aux = aux.prox
    if cont > maiorCont:
      maiorCont = cont
      maior = head.valor
    head = head.prox
  return maior, maiorCont
#B) Complexidade: O(n^2) ja que temos whiles aninhados
```

```python
#Questão 9
def temRepetido(p):
  head = p
  while head is not None:
    aux = head.prox
    while aux is not None:
      if head.valor == aux.valor:
        return True
      aux = aux.prox
    head = head.prox
  return False
#B)Complexidade: O(n^2), por ter dois whiles aninhados
```

```python
#Questão 10
def intersecao(p1, p2):
  p = None
  fim = None
  head = p1
  while head is not None:
    aux = p2
    while aux is not None:
      if head.valor == aux.valor:
        novo = No(head.valor)
        if p is None:
          p = novo
          fim = novo
        else:
          fim.prox = novo
          fim = nov
        break
      aux = aux.prox
    head = head.prox
  return p

#B)Complexidade : O(n^2)
```
