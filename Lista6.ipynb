```python
#Tipo abstrato de dados
class No:
  def __init__(self, valor):
    self.valor = valor
    self.prox = None
```

```python
#Questão1
#Tad da Pilha em listas encadeadas
class Pilha:
  def __init__(self):
      self.topo = None

  def push(self, valor):
      novo = No(valor)
      novo.prox = self.topo
      self.topo = novo

  def pop(self):
      if self.topo is None:
          return None
      temp = self.topo
      valor = temp.valor
      self.topo = temp.prox
      temp.prox = None  # desliga o no removido da pilha
      return valor

  def vazia(self):
      return self.topo is None
#A)o estado da pilha fica topo->3->99->42->17->5->null

#B)Os valores removidos serão 3 e 99, a pilha ficará:
# topo->42->17->5->null

#C)Precisamos guardar o nó a ser removido porque teremos que acessar sua informação no caso o valor
```

```python
#Questão 2
#Tad da fila em listas encadeadas
class Fila:
  def __init__(self):
      self.inicio = None
      self.fim = None

  def enqueue(self, valor):
    novo = No(valor)
    if self.inicio is None:
      self.inicio = novo
      self.fim = novo
    else:
      self.fim.prox = novo
      self.fim = novo
  def dequeue(self):
    if self.inicio is None:
      return None
    temp = self.inicio
    valor = temp.valor
    self.inicio = temp.prox
    if self.inicio is None:
      self.fim = None
    temp.prox = None
    return valor

  def vazia(self):
    return self.inicio is None
#A)Inserido 7 e 11 na fila temos:
# inicio -> 8 ->15 -> 23 -> 7-> 11 <- fim

#B) Realizando dois dequeue teremos:
# inicio -> 23-> 7 -> 11 <- fim

#C)Como nesse caso em específico a lista fica vazia os dois ponteiros
# devem ser atualizados para None
```

```python
#Questão 3
#A)
def inverterFilaComPilha(fila):
  pilha = Pilha()
  while not fila.vazia():
    pilha.push(fila.dequeue())
  while not pilha.vazia():
    fila.enqueue(pilha.pop())
  return fila

#B) A complexidade desse algoritmo fica o(n)
# ja que cada elemento pasa pela pilha uma vez em ambos
# os whiles

#C) Sim é possível inverter a fila sem estruturas auxiliares
# basta utilizar o algoritmo basico de inversões de listas encadeadss
# invertendo o sentindo do ponteiro paro o prox de cada nó
```

```python
#Questão 5
#A)
def balanceado(expressao):
  pilha = Pilha()
  pares = {')': '(', ']': '[', '}': '{'}
  abridores = {'(', '[', '{'}
  for i, c in enumerate(expressao):
    if c in abridores:
      pilha.push(c)
    elif c in pares:
      if pilha.vazia():
        return False, i, c
      topoAtual = pilha.pop()
      if topoAtual != pares[c]:
        return False, i, c

  if not pilha.vazia():
      return False, len(expressao), None  # sobrou abridor sem fechar
  return True, -1, None

#B) Complexidade: O(n), com n sendo o tamanho da expressão

#C)A primeira é válida
# A segunda é Inválidao erro foi na posição 3
# A terceira é válida
```

```python
#Questão 6
class FilaDuasPilhas:
  def __init__(self):
    self.p1 = Pilha()  # entrada
    self.p2 = Pilha()  # saida
  def enqueue(self, valor):
    self.p1.push(valor)

  def dequeue(self):
    if self.p2.vazia():
      while not self.p1.vazia():
        self.p2.push(self.p1.pop())
    return self.p2.pop()

#B)Cada elemento é empurrado para p1 uma vez, transferido
# de p1 para p2 no maximo uma vez, ee retirado de p2 uma vez.Logo
# o custo é de n operações O(n), o que da um custo amortizado
# por dequeue

#C) Sim preserva o FIFO. Exemplo: enqueue(1), enqueue(2),
# enqueue(3) p1 fica com topo=3,2,1 com o 3 no topo.
# Ao transferir tudo par p2, a ordem se inverte de novo:
# p2 fica com 1,2,3 com 1 no topo
```
