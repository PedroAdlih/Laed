```python
# Questão 4
def repetido(M):
  n = len(M)
  for i1 in range(n):
    for j1 in range(n):
      for i2 in range(n):
        for j2 in range(n):
          if i1 == i2 and j1 == j2:
            continue
          if M[i1][j1] == M[i2][j2]:
           return True
  return False
M = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 4, 9]
]
if repetido(M):
    print("Elemento Repetido")
else:
    print("Não existem elementos repetidos.")
```

```python
# Questão 5
def linhas_iguais(M):
  n = len(M)
  for i in range(n):
    for j in range(i + 1, n):
      iguais = True
      for k in range(len(M[i])):
        if M[i][k] != M[j][k]:
          iguais = False
          break
      if iguais:
        return True
  return False
M = [
    [1, 2, 3],
    [4, 5, 6],
    [1, 2, 3]
]
if linhas_iguais(M):
    print("Sim, existem duas linhas iguais.")
else:
    print("Não existem duas linhas iguais.")
```
